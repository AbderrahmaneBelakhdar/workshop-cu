```python
# Load CSV file
csv_name = 'cars.csv'

try:
    from google.colab import drive
    drive.mount('/content/drive')
except Exception:
    pass

search_roots = [Path('.'), Path('/content/drive/MyDrive')]
candidates = []
for root in search_roots:
    if root.exists():
        candidates.extend(root.rglob(csv_name))

if not candidates:
    raise FileNotFoundError(f"{csv_name} not found. Put it in the notebook folder or in Google Drive.")

csv_path = candidates[0]
print(f"CSV loaded from: {csv_path}")

df_csv = pd.read_csv(csv_path)
df = df_csv.copy()
```

```python
import pandas as pd

df = pd.read_csv("cars.csv")
```

```python
print("Columns:\n", df.columns)

print("\nData Types:\n", df.dtypes)

print("\nShape (rows, columns):", df.shape)
```

```python
print("Number of unique fuel types:", df['fuel_type'].nunique())

print("\nFuel distribution:")
print(df['fuel_type'].value_counts())
```

```python
import numpy as np

df['engine'] = df['engine'].replace(0, np.nan)
df['price_per_kg'] = df['price'] / df['engine']

top10 = df.sort_values(by='price', ascending=False).head(10)

print(top10[['car_name', 'brand', 'price']])
```
